```polyglot-notebook

using System;
using System.Collections.Generic;

// ========================================
// БАЗОВЫЙ КЛАСС Plant
// ========================================
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    // Метод цветения
    public virtual void Цветение()
    {
        Console.WriteLine($"{Name} начинает цвести.");
    }

    // Метод роста
    public void Растет(float рост)
    {
        Height += рост;

        Console.WriteLine(
            $"{Name} вырос на {рост} см. Новая высота: {Height} см."
        );
    }

    // Метод полива
    public virtual void Полив()
    {
        Console.WriteLine(
            $"{Name} получил воду и чувствует себя хорошо."
        );
    }

    // Вывод информации
    public virtual void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Высота: {Height} см");
    }
}


// ========================================
// КЛАСС Flower
// ========================================
public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public string Состояние { get; private set; }

    public Flower(
        string name,
        float height,
        string color,
        int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
        Состояние = "Цветёт";
    }

    // Переопределение метода цветения
    public override void Цветение()
    {
        Console.WriteLine(
            $"{Name} распускает свои красивые цветы!"
        );

        Console.WriteLine(
            $"Цвет: {Color}, количество лепестков: {PetalCount}."
        );
    }

    // Метод увядания
    public void Увядать()
    {
        Состояние = "Увядает";

        Console.WriteLine($"{Name} начинает увядать.");
        Console.WriteLine($"Текущее состояние: {Состояние}");
    }

    // Переопределение метода полива
    public override void Полив()
    {
        if (Name == "Орхидея")
        {
            Console.WriteLine(
                "Орхидея напилась воды и раскрыла свои нежные цветы."
            );
        }
        else
        {
            Console.WriteLine(
                $"{Name} получил воду и продолжает радовать своей красотой."
            );
        }
    }

    // Информация о цветке
    public override void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Цвет: {Color}");
        Console.WriteLine($"Количество лепестков: {PetalCount}");
        Console.WriteLine($"Высота: {Height} см");
        Console.WriteLine($"Состояние: {Состояние}");
    }
}


// ========================================
// КЛАСС Tree
// ========================================
public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(
        string name,
        float height,
        int age,
        string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    // Переопределение метода цветения
    public override void Цветение()
    {
        Console.WriteLine(
            $"{Name} растёт и покрывается свежей листвой."
        );

        Console.WriteLine(
            $"Возраст: {Age} лет, тип листьев: {LeafType}."
        );
    }

    // Увеличение возраста дерева
    public void Вырастить(int годы)
    {
        Age += годы;

        Console.WriteLine(
            $"{Name} стал старше на {годы} лет."
        );

        Console.WriteLine(
            $"Теперь возраст дерева: {Age} лет."
        );
    }

    // Полив дерева
    public override void Полив()
    {
        Console.WriteLine(
            $"{Name} получил воду и укрепляет свои корни."
        );
    }

    // Информация о дереве
    public override void ПоказатьИнформацию()
    {
        Console.WriteLine($"Название: {Name}");
        Console.WriteLine($"Тип: {Type}");
        Console.WriteLine($"Высота: {Height} см");
        Console.WriteLine($"Возраст: {Age} лет");
        Console.WriteLine($"Тип листьев: {LeafType}");
    }
}


// ========================================
// КЛАСС Garden
// ========================================
public class Garden
{
    public string Name { get; set; }

    private List<Plant> Plants { get; set; }

    public Garden(string name)
    {
        Name = name;
        Plants = new List<Plant>();
    }

    // Добавление растения
    public void ДобавитьРастение(Plant plant)
    {
        Plants.Add(plant);
    }

    // Показать все растения
    public void ПоказатьСад()
    {
        Console.WriteLine();
        Console.WriteLine("========================================");
        Console.WriteLine($"       {Name}");
        Console.WriteLine("========================================");

        foreach (Plant plant in Plants)
        {
            Console.WriteLine();
            Console.WriteLine("----------------------------------------");

            plant.ПоказатьИнформацию();

            Console.WriteLine("----------------------------------------");
        }
    }

    // Полиморфный вызов цветения
    public void ВсеЦветут()
    {
        Console.WriteLine();
        Console.WriteLine("============ ЦВЕТЕНИЕ ==================");

        foreach (Plant plant in Plants)
        {
            plant.Цветение();
        }
    }

    // Полив всех растений
    public void ПолитьВсе()
    {
        Console.WriteLine();
        Console.WriteLine("============ ПОЛИВ =====================");

        foreach (Plant plant in Plants)
        {
            plant.Полив();
        }
    }
}


// ========================================
// СОЗДАНИЕ САДА НГАРГОНДО ТИДЖАНИ
// ========================================

Garden сад = new Garden("САД НГАРГОНДО ТИДЖАНИ");


// ========================================
// СОЗДАНИЕ ЦВЕТОВ
// ========================================

Flower орхидея = new Flower(
    "Орхидея",
    45,
    "Фиолетовая",
    6
);

Flower лилия = new Flower(
    "Лилия",
    65,
    "Белая",
    12
);


// ========================================
// СОЗДАНИЕ ДЕРЕВЬЕВ
// ========================================

Tree манго = new Tree(
    "Манго",
    280,
    10,
    "Лиственные"
);

Tree баобаб = new Tree(
    "Баобаб",
    450,
    25,
    "Лиственные"
);


// ========================================
// ДОБАВЛЕНИЕ РАСТЕНИЙ В САД
// ========================================

сад.ДобавитьРастение(орхидея);
сад.ДобавитьРастение(лилия);
сад.ДобавитьРастение(манго);
сад.ДобавитьРастение(баобаб);


// ========================================
// НАЧАЛЬНОЕ СОСТОЯНИЕ САДА
// ========================================

Console.WriteLine("НАЧАЛЬНОЕ СОСТОЯНИЕ САДА");

сад.ПоказатьСад();


// ========================================
// ЦВЕТЕНИЕ — ПОЛИМОРФИЗМ
// ========================================

сад.ВсеЦветут();


// ========================================
// РОСТ РАСТЕНИЙ
// ========================================

Console.WriteLine();
Console.WriteLine("============ РОСТ РАСТЕНИЙ =============");

орхидея.Растет(8);
лилия.Растет(5);
манго.Растет(15);
баобаб.Растет(12);


// ========================================
// УВЕЛИЧЕНИЕ ВОЗРАСТА ДЕРЕВЬЕВ
// ========================================

Console.WriteLine();
Console.WriteLine("============ ВОЗРАСТ ДЕРЕВЬЕВ ==========");

манго.Вырастить(2);
баобаб.Вырастить(3);


// ========================================
// ПОЛИВ ВСЕХ РАСТЕНИЙ
// ========================================

сад.ПолитьВсе();


// ========================================
// УВЯДАНИЕ ЛИЛИИ
// ========================================

Console.WriteLine();
Console.WriteLine("============ УВЯДАНИЕ ==================");

лилия.Увядать();


// ========================================
// ФИНАЛЬНОЕ СОСТОЯНИЕ САДА
// ========================================

Console.WriteLine();
Console.WriteLine("============ ИТОГОВОЕ СОСТОЯНИЕ =========");

сад.ПоказатьСад();


// ========================================
// КОНЕЦ ПРОГРАММЫ
// ========================================

Console.WriteLine();
Console.WriteLine("Программа завершена.");
```
