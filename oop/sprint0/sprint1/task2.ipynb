```polyglot-notebook
using System;

// Базовый класс Animal
public class Animal
{
    private int _age;

    // Свойства с геттером и сеттером
    public string Name { get; set; }

    public int Age
    {
        get { return _age; }
        set
        {
            if (value >= 0)
                _age = value;
            else
                throw new ArgumentOutOfRangeException(
                    "Возраст не может быть отрицательным!");
        }
    }

    public string Color { get; set; }

    // Статический атрибут
    public static int TotalAnimals { get; private set; }

    // Конструктор
    public Animal(string name, int age, string color)
    {
        Name = name;
        Age = age;
        Color = color;

        TotalAnimals++;
    }

    // Обычный метод
    public void Eat()
    {
        Console.WriteLine($"{Name} ест.");
    }

    // Обычный метод
    public void Sleep()
    {
        Console.WriteLine($"{Name} спит.");
    }

    // Статический метод
    public static void ShowTotalAnimals()
    {
        Console.WriteLine($"Всего животных: {TotalAnimals}");
    }
}


// Производный класс Dog
public class Dog : Animal
{
    public string Breed { get; set; }

    public Dog(string name, int age, string color, string breed)
        : base(name, age, color)
    {
        Breed = breed;
    }

    public void Bark()
    {
        Console.WriteLine($"{Name}: Гав-гав!");
    }

    public void PlayWith(Cat cat)
    {
        Console.WriteLine($"{Name} играет с котом {cat.Name}.");
    }
}


// Производный класс Cat
public class Cat : Animal
{
    public bool IsDomestic { get; set; }

    public Cat(string name, int age, string color, bool isDomestic)
        : base(name, age, color)
    {
        IsDomestic = isDomestic;
    }

    public void Meow()
    {
        Console.WriteLine($"{Name}: Мяу!");
    }

    public void PlayWith(Dog dog)
    {
        Console.WriteLine($"{Name} играет с собакой {dog.Name}.");
    }
}


// Класс Owner для демонстрации взаимодействия объектов
public class Owner
{
    public string Name { get; set; }

    public Owner(string name)
    {
        Name = name;
    }

    // Взаимодействие владельца с животным
    public void FeedAnimal(Animal animal)
    {
        Console.WriteLine($"{Name} кормит {animal.Name}.");
        animal.Eat();
    }

    public void PlayWithAnimal(Animal animal)
    {
        Console.WriteLine($"{Name} играет с {animal.Name}.");
    }
}


// Создание объектов
Dog dog = new Dog(
    "Шарик",
    3,
    "Коричневый",
    "Овчарка"
);

Cat cat = new Cat(
    "Барсик",
    2,
    "Серый",
    true
);

Owner owner = new Owner("Игорь");


// Вывод информации о собаке
Console.WriteLine("=== Собака ===");
Console.WriteLine($"Имя: {dog.Name}");
Console.WriteLine($"Возраст: {dog.Age}");
Console.WriteLine($"Цвет: {dog.Color}");
Console.WriteLine($"Порода: {dog.Breed}");


// Вывод информации о кошке
Console.WriteLine();
Console.WriteLine("=== Кошка ===");
Console.WriteLine($"Имя: {cat.Name}");
Console.WriteLine($"Возраст: {cat.Age}");
Console.WriteLine($"Цвет: {cat.Color}");
Console.WriteLine($"Домашняя: {cat.IsDomestic}");


// Взаимодействие владельца с животными
Console.WriteLine();
Console.WriteLine("=== Взаимодействие объектов ===");

owner.FeedAnimal(dog);
owner.FeedAnimal(cat);

owner.PlayWithAnimal(dog);
owner.PlayWithAnimal(cat);


// Взаимодействие животных между собой
dog.PlayWith(cat);
cat.PlayWith(dog);

dog.Bark();
cat.Meow();


// Изменение свойств объектов через сеттер
Console.WriteLine();
Console.WriteLine("=== Изменение свойств ===");

dog.Age = 4;
dog.Color = "Чёрный";

Console.WriteLine(
    $"Собака {dog.Name}: возраст {dog.Age}, цвет {dog.Color}"
);


// Обычные методы
dog.Sleep();
cat.Sleep();


// Статический метод
Console.WriteLine();
Animal.ShowTotalAnimals();
```
