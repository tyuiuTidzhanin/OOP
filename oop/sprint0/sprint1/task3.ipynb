```polyglot-notebook
using System;
using System.Collections.Generic;

// ========================================
// КЛАСС PRODUCT — ТОВАР
// ========================================
public class Product
{
    private string name;
    private decimal price;
    private int quantity;

    protected string category;

    internal string description;

    // Геттер и сеттер Name
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    // Геттер и сеттер Price
    public decimal Price
    {
        get { return price; }
        set
        {
            if (value >= 0)
                price = value;
        }
    }

    // Геттер и сеттер Quantity
    public int Quantity
    {
        get { return quantity; }
        set
        {
            if (value >= 0)
                quantity = value;
        }
    }

    public Product(
        string name,
        decimal price,
        int quantity,
        string category,
        string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    // Общая стоимость всех единиц товара
    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    // Изменение количества товара
    public void UpdateQuantity(int amount)
    {
        quantity += amount;

        if (quantity < 0)
            quantity = 0;

        Console.WriteLine(
            $"Количество товара «{name}» изменено. " +
            $"Теперь на складе: {quantity} шт."
        );
    }

    // Информация о товаре
    public string GetProductInfo()
    {
        return
            $"Товар: {name}\n" +
            $"Цена: {price} ₽\n" +
            $"Количество: {quantity} шт.\n" +
            $"Категория: {category}\n" +
            $"Описание: {description}\n" +
            $"Общая стоимость: {GetTotalPrice()} ₽";
    }
}


// ========================================
// КЛАСС SELLER — ПРОДАВЕЦ
// ========================================
public class Seller
{
    private string name;
    private string employeeId;

    protected decimal salary;

    internal string contactInfo;

    private List<Product> products;

    // Геттер и сеттер Name
    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    // Геттер и сеттер EmployeeId
    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }

    public Seller(
        string name,
        string employeeId,
        decimal salary,
        string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;

        products = new List<Product>();
    }

    // Добавление товара продавцу
    public void AddProduct(Product product)
    {
        products.Add(product);

        Console.WriteLine(
            $"Продавец {name} добавил товар «{product.Name}»."
        );
    }

    // Продажа товара
    public void SellProduct(Product product, int quantity)
    {
        if (quantity <= 0)
        {
            Console.WriteLine(
                "Количество для продажи должно быть больше 0."
            );
            return;
        }

        if (quantity > product.Quantity)
        {
            Console.WriteLine(
                $"Нельзя продать {quantity} шт. " +
                $"На складе только {product.Quantity} шт."
            );
            return;
        }

        product.UpdateQuantity(-quantity);

        Console.WriteLine(
            $"Продавец {name} продал {quantity} шт. " +
            $"товара «{product.Name}»."
        );
    }

    // Информация о продавце
    public string GetSellerInfo()
    {
        return
            $"Продавец: {name}\n" +
            $"Табельный номер: {employeeId}\n" +
            $"Зарплата: {salary} ₽\n" +
            $"Контакты: {contactInfo}\n" +
            $"Товаров у продавца: {products.Count}";
    }
}


// ========================================
// КЛАСС STORE — МАГАЗИН
// ========================================
public class Store
{
    private string storeName;
    private string location;

    public string storeHours;

    private List<Seller> sellers;
    private List<Product> products;

    // Геттер и сеттер StoreName
    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    // Геттер и сеттер Location
    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public Store(
        string storeName,
        string location,
        string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;

        sellers = new List<Seller>();
        products = new List<Product>();
    }

    // Добавление продавца
    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);

        Console.WriteLine(
            $"Продавец {seller.Name} добавлен в магазин."
        );
    }

    // Добавление товара
    public void AddProduct(Product product)
    {
        products.Add(product);

        Console.WriteLine(
            $"Товар «{product.Name}» добавлен в магазин."
        );
    }

    // Список товаров
    public void ListProducts()
    {
        Console.WriteLine();
        Console.WriteLine("============ ТОВАРЫ МАГАЗИНА ============");

        foreach (Product product in products)
        {
            Console.WriteLine();
            Console.WriteLine(product.GetProductInfo());
            Console.WriteLine("------------------------------------------");
        }
    }

    // Информация о магазине
    public string GetStoreInfo()
    {
        return
            $"Магазин: {storeName}\n" +
            $"Адрес: {location}\n" +
            $"Режим работы: {storeHours}\n" +
            $"Продавцов: {sellers.Count}\n" +
            $"Товаров: {products.Count}";
    }
}


// ========================================
// СОЗДАНИЕ ТОВАРА
// ========================================

Product энергетик = new Product(
    "Энергетик",
    150m,
    3,
    "Напитки",
    "Газированный энергетический напиток"
);


// ========================================
// СОЗДАНИЕ ПРОДАВЦА
// ========================================

Seller артем = new Seller(
    "Артём",
    "67",
    45000m,
    "+7 900 456-78-12"
);


// ========================================
// СОЗДАНИЕ МАГАЗИНА
// ========================================

Store магнит = new Store(
    "Магнит",
    "Олимпийская, 9",
    "Круглосуточно"
);


// ========================================
// ДОБАВЛЕНИЕ ТОВАРА И ПРОДАВЦА
// ========================================

магнит.AddProduct(энергетик);

артем.AddProduct(энергетик);

магнит.AddSeller(артем);


// ========================================
// ИНФОРМАЦИЯ О МАГАЗИНЕ
// ========================================

Console.WriteLine();
Console.WriteLine("========================================");
Console.WriteLine("          ИНФОРМАЦИЯ О МАГАЗИНЕ");
Console.WriteLine("========================================");

Console.WriteLine(магнит.GetStoreInfo());


// ========================================
// ИНФОРМАЦИЯ О ПРОДАВЦЕ
// ========================================

Console.WriteLine();
Console.WriteLine("========================================");
Console.WriteLine("          ИНФОРМАЦИЯ О ПРОДАВЦЕ");
Console.WriteLine("========================================");

Console.WriteLine(артем.GetSellerInfo());


// ========================================
// ИНФОРМАЦИЯ О ТОВАРЕ
// ========================================

Console.WriteLine();
Console.WriteLine("========================================");
Console.WriteLine("             ИНФОРМАЦИЯ О ТОВАРЕ");
Console.WriteLine("========================================");

Console.WriteLine(энергетик.GetProductInfo());


// ========================================
// СПИСОК ТОВАРОВ
// ========================================

магнит.ListProducts();


// ========================================
// ПРОДАЖА ТОВАРА
// ========================================

Console.WriteLine();
Console.WriteLine("============ ПРОДАЖА ===================");

артем.SellProduct(энергетик, 1);


// ========================================
// ИТОГ
// ========================================

Console.WriteLine();
Console.WriteLine("============ ПОСЛЕ ПРОДАЖИ =============");

Console.WriteLine($"Товар: {энергетик.Name}");
Console.WriteLine($"Цена: {энергетик.Price} ₽");
Console.WriteLine($"Осталось: {энергетик.Quantity} шт.");
Console.WriteLine(
    $"Общая стоимость оставшегося товара: {энергетик.GetTotalPrice()} ₽"
);

Console.WriteLine();
Console.WriteLine("========================================");
Console.WriteLine("           ПРОГРАММА ЗАВЕРШЕНА");
Console.WriteLine("========================================");
```
