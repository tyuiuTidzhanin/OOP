```polyglot-notebook

using System;

// Интерфейс
public interface IAnimal
{
    void MakeSound();
}

// Базовый класс Animal
public class Animal : IAnimal
{
    // Свойства
    public string Name { get; set; }
    public int Age { get; set; }

    // Приватное поле
    private int _energy;

    // Protected поле
    protected string Species;

    // Static член класса
    public static int TotalAnimals { get; private set; }

    // Internal член
    internal string Description { get; set; }

    // Конструктор по умолчанию
    public Animal()
    {
        Name = "Тиджани";
        Age = 20;
        _energy = 100;
        Species = "Неизвестный вид";
        Description = "Животное";

        TotalAnimals++;
    }

    // Конструктор с параметрами
    public Animal(string name, int age, string species)
    {
        Name = name;
        Age = age;
        _energy = 100;
        Species = species;
        Description = "Животное";

        TotalAnimals++;
    }

    // Метод кормления животного
    public void Eat(int food)
    {
        _energy += food;

        if (_energy > 100)
            _energy = 100;

        Console.WriteLine($"{Name} поело. Энергия: {_energy}");
    }

    // Метод движения животного
    public void Move(int energy)
    {
        _energy -= energy;

        if (_energy < 0)
            _energy = 0;

        Console.WriteLine($"{Name} двигается. Энергия: {_energy}");
    }

    // Реализация интерфейса
    public void MakeSound()
    {
        Console.WriteLine($"{Name} издаёт звук");
    }

    // Protected метод
    protected void ShowSpecies()
    {
        Console.WriteLine($"Вид: {Species}");
    }

    // Static метод
    public static void ShowTotalAnimals()
    {
        Console.WriteLine($"Всего создано животных: {TotalAnimals}");
    }

    // Метод отображения информации
    public void ShowInfo()
    {
        Console.WriteLine("----- Информация о животном -----");
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Возраст: {Age}");
        Console.WriteLine($"Вид: {Species}");
        Console.WriteLine($"Энергия: {_energy}");
        Console.WriteLine($"Описание: {Description}");
    }
}

// Производный класс Dog
public class Dog : Animal
{
    public Dog(string name, int age)
        : base(name, age, "Собака")
    {
    }

    // Метод собачьего лая
    public void Bark()
    {
        Console.WriteLine($"{Name} говорит: Гав-гав!");
    }

    // Доступ к protected методу
    public void ShowDogSpecies()
    {
        ShowSpecies();
    }
}

// Создание объекта с помощью конструктора с параметрами
Animal animal = new Animal("Тиджани", 20, "Кот");

animal.ShowInfo();
animal.MakeSound();
animal.Eat(20);
animal.Move(30);

Console.WriteLine();

// Создание объекта производного класса
Dog dog = new Dog("Рекс", 4);

dog.ShowInfo();
dog.Bark();
dog.ShowDogSpecies();

Console.WriteLine();

// Вызов статического метода
Animal.ShowTotalAnimals();
```
