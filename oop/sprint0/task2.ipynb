```polyglot-notebook
public class Animal
{
    public virtual void Move()
    {
        Console.WriteLine("Животное движется");
    }
}

public interface ISoundable
{
    void MakeSound();
}

public class Cat : Animal, ISoundable
{
    public void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

Cat myCat = new Cat();

myCat.Move();
myCat.MakeSound();
```

```polyglot-notebook
public interface IMovable
{
    void Move();
}

public class Animal : IMovable
{
    public void Move()
    {
        Console.WriteLine("Животное движется");
    }
}

public class Cat : IMovable
{
    public void Move()
    {
        Console.WriteLine("Кот бегает");
    }

    public void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myAnimal.Move();
myCat.Move();
myCat.MakeSound();
```

```polyglot-notebook
public interface ILandAnimal
{
    void Move();
}

public interface IAirAnimal
{
    void Fly();
}

public class Cat : ILandAnimal
{
    public void Move()
    {
        Console.WriteLine("Кот движется");
    }

    public void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

public class Bird : IAirAnimal
{
    public void Fly()
    {
        Console.WriteLine("Птица летает");
    }
}

Cat myCat = new Cat();
Bird myBird = new Bird();

myCat.Move();
myBird.Fly();
```

```polyglot-notebook
public interface IAnimal
{
    void Move();
}

public interface ICanMakeSound
{
    void MakeSound();
}

public interface ICanFly
{
    void Fly();
}

public interface ICanHonk
{
    void Honk();
}

public class Cat : IAnimal, ICanMakeSound
{
    public void Move()
    {
        Console.WriteLine("Кот движется");
    }

    public void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

public class Bird : IAnimal, ICanFly
{
    public void Move()
    {
        Console.WriteLine("Птица движется");
    }

    public void Fly()
    {
        Console.WriteLine("Птица летит");
    }
}

Cat myCat = new Cat();
Bird myBird = new Bird();

myCat.Move();
myCat.MakeSound();

myBird.Move();
myBird.Fly();
```

```polyglot-notebook
public interface IFood
{
    void Eat();
}

public class Grass : IFood
{
    public void Eat()
    {
        Console.WriteLine("Животное ест траву");
    }
}

public class Animal
{
    private IFood _food;

    public Animal(IFood food)
    {
        _food = food;
    }

    public void EatFood()
    {
        _food.Eat();
    }
}

IFood food = new Grass();
Animal myAnimal = new Animal(food);

myAnimal.EatFood();
```
