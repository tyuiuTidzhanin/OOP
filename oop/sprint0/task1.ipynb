```polyglot-notebook
// Создание класса
class Animal
{
    private string color; // Приватное поле

    public string Color // Свойство с геттером и сеттером
    {
        get { return color; }
        set { color = value; }
    }

    public void Move() // Публичный метод
    {
        Console.WriteLine("Животное движется");
    }
}

// Основная программа
Animal MyAnimal = new Animal(); // Создание объекта
MyAnimal.Color = "Рыжий";
MyAnimal.Move();
Console.WriteLine(MyAnimal.Move);
```

```polyglot-notebook
class Animal
{
    private string color;

    public string Color
    {
        get { return color; }
        set { color = value; }
    }

    public void Sit()
    {
        Console.WriteLine("Животное сидит");
    }
}

class Cat : Animal
{
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myCat.Color = "Рыжий";
Console.WriteLine(myCat.Color);
myCat.Sit();

myAnimal.Color = "Красный";
Console.WriteLine(myAnimal.Color);
myAnimal.Sit();
```

```polyglot-notebook
class Animal
{
    public virtual void MakeSound()
    {
        Console.WriteLine("Животное издает звук");
    }
}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

Animal myAnimal = new Animal();
Cat myCat = new Cat();

myAnimal.MakeSound();
myCat.MakeSound();
```

```polyglot-notebook
abstract class Animal
{
    public abstract void MakeSound();

    public void Move()
    {
        Console.WriteLine("Животное движется");
    }
}

class Cat : Animal
{
    public override void MakeSound()
    {
        Console.WriteLine("Кот мяукает");
    }
}

Cat myCat = new Cat();

myCat.MakeSound();
myCat.Move();
```

```polyglot-notebook
class Heart
{
    public void Beat()
    {
        Console.WriteLine("Сердце бьется");
    }
}

class Animal
{
    private Heart heart;

    public Animal(Heart heart)
    {
        this.heart = heart;
    }

    public void Live()
    {
        heart.Beat();
        Console.WriteLine("Животное в порядке");
    }
}

Heart myHeart = new Heart();
Animal myAnimal = new Animal(myHeart);

myAnimal.Live();
```

```polyglot-notebook
abstract class LivingBeing
{
    public abstract void Live();
}

class Animal : LivingBeing
{
    private Owner owner;

    public Animal(Owner owner)
    {
        this.owner = owner;
    }

    public override void Live()
    {
        Console.WriteLine("Животное в порядке");
    }
}

class Owner
{
    public string Name { get; set; }

    public Owner(string name)
    {
        Name = name;
    }

    public void Care(LivingBeing livingBeing)
    {
        Console.WriteLine($"{Name} заботится о животном...");
        livingBeing.Live();
    }
}

Owner Ivan = new Owner("Иван");
Animal myAnimal = new Animal(Ivan);

myAnimal.Live();
Ivan.Care(myAnimal);
```
