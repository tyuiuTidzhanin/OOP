```polyglot-notebook
int[] numbers = { 5, 8, 1, 3, 9, 6, 7, 2 };

int min = numbers.Min();
int max = numbers.Max();

Console.WriteLine($"Минимальное значение: {min}");
Console.WriteLine($"Максимальное значение: {max}");

int sum = numbers.Sum();

Console.WriteLine($"Сумма элементов: {sum}");

double average = numbers.Average();

Console.WriteLine($"Среднее арифметическое: {average}");

int[] sortedNumbers = numbers.OrderBy(x => x).ToArray();

Console.WriteLine("Массив по возрастанию:");
Console.WriteLine(string.Join(", ", sortedNumbers));

int numberToFind = 6;
bool contains = numbers.Contains(numberToFind);

Console.WriteLine($"Содержит ли массив число {numberToFind}: {contains}");
```
